```python

import pandas as pd
import numpy as np
from pathlib import Path

# ============================================================
# 1. LOCALIZAR E CARREGAR AS BASES
# ============================================================

diretorio = Path.cwd()

arquivos = [
    arq for arq in diretorio.rglob("*.csv")
    if not any(
        p in arq.parts
        for p in [".git", ".venv", "venv", "site-packages",
                  "node_modules", ".ipynb_checkpoints"]
    )
]

bases = {}

# DataFrames já carregados no notebook
for nome, objeto in list(globals().items()):
    if isinstance(objeto, pd.DataFrame) and not objeto.empty:
        bases[f"DataFrame: {nome}"] = objeto

# Arquivos CSV encontrados
for arquivo in arquivos:
    try:
        df = pd.read_csv(arquivo, low_memory=False, encoding="utf-8")
    except UnicodeDecodeError:
        df = pd.read_csv(arquivo, low_memory=False, encoding="latin1")

    bases[f"Arquivo: {arquivo.name}"] = df

print(f"Total de bases identificadas: {len(bases)}")

# ============================================================
# 2. INVENTÁRIO DE COLUNAS
# ============================================================

inventario_colunas = []
inventario_valores = []
resumo_bases = []

for nome_base, df in bases.items():

    print("\n" + "=" * 100)
    print(f"BASE: {nome_base}")
    print("=" * 100)

    contagem_tipos = {
        "Numéricas": 0,
        "Categóricas": 0,
        "Textuais": 0,
        "Datas": 0,
        "Outras": 0
    }

    for coluna in df.columns:

        serie = df[coluna]
        tipo = str(serie.dtype)

        # Classificação dos tipos
        if pd.api.types.is_datetime64_any_dtype(serie):
            classificacao = "Datas"
        elif pd.api.types.is_numeric_dtype(serie):
            classificacao = "Numéricas"
        elif isinstance(serie.dtype, pd.CategoricalDtype):
            classificacao = "Categóricas"
        elif pd.api.types.is_string_dtype(serie) or serie.dtype == object:
            classificacao = "Textuais"
        else:
            classificacao = "Outras"

        contagem_tipos[classificacao] += 1

        ausentes = serie.isna().sum()
        distintos = serie.nunique(dropna=True)

        duplicados = serie.duplicated().sum()

        # Registrar informações da coluna
        inventario_colunas.append({
            "base": nome_base,
            "coluna": coluna,
            "tipo_original": tipo,
            "classificacao": classificacao,
            "total_linhas": len(df),
            "valores_preenchidos": serie.count(),
            "valores_ausentes": ausentes,
            "percentual_ausente": round(
                ausentes / len(df) * 100, 2
            ) if len(df) else 0,
            "valores_distintos": distintos,
            "registros_duplicados": duplicados
        })

        print(
            f"{coluna} | Tipo: {tipo} | "
            f"Classificação: {classificacao} | "
            f"Ausentes: {ausentes:,} | "
            f"Distintos: {distintos:,}"
        )

        # ====================================================
        # 3. LEVANTAR VALORES TEXTUAIS E CATEGÓRICOS
        # ====================================================

        if classificacao in ["Textuais", "Categóricas"]:

            valores = serie.dropna().astype(str).str.strip()
            valores = valores[valores != ""]

            frequencias = valores.value_counts()

            for valor, frequencia in frequencias.items():

                inventario_valores.append({
                    "base": nome_base,
                    "coluna": coluna,
                    "valor": valor,
                    "frequencia": frequencia
                })

    # Resumo da base
    resumo_bases.append({
        "base": nome_base,
        "linhas": len(df),
        "total_colunas": len(df.columns),
        **contagem_tipos
    })

# ============================================================
# 4. CRIAR DATAFRAMES DOS INVENTÁRIOS
# ============================================================

df_colunas = pd.DataFrame(inventario_colunas)
df_valores = pd.DataFrame(inventario_valores)
df_resumo = pd.DataFrame(resumo_bases)

# ============================================================
# 5. EXPORTAR RESULTADOS
# ============================================================

df_colunas.to_csv(
    "inventario_completo_colunas.csv",
    index=False,
    encoding="utf-8-sig"
)

df_valores.to_csv(
    "inventario_valores_categoricos.csv",
    index=False,
    encoding="utf-8-sig"
)

df_resumo.to_csv(
    "resumo_geral_bases.csv",
    index=False,
    encoding="utf-8-sig"
)

# ============================================================
# 6. EXIBIR RESUMOS
# ============================================================

print("\n" + "=" * 100)
print("RESUMO GERAL DAS BASES")
print("=" * 100)

display(df_resumo)

print("\n" + "=" * 100)
print("INVENTÁRIO COMPLETO DE COLUNAS")
print("=" * 100)

display(df_colunas)

print("\n" + "=" * 100)
print("INVENTÁRIO DE VALORES CATEGÓRICOS E TEXTUAIS")
print("=" * 100)

display(df_valores)

print("\nArquivos exportados:")
print("1. inventario_completo_colunas.csv")
print("2. inventario_valores_categoricos.csv")
print("3. resumo_geral_bases.csv")
```
